```python
import tensorflow as tf
import matplotlib.pyplot as plt

# 画像の水増し関数（左右反転）
def flip_left_right(image, label):
    image = tf.image.flip_left_right(image)
    return image, label
    return image, label

# 画像の水増し関数（反時計回りに180度回転）
def rot180(image, label):
    image = tf.image.rot90(image, k=2)
    return image, label
    return image, label

# 訓練データ画像読込
train_dataset = tf.keras.preprocessing.image_dataset_from_directory(
    "dog_cat_photos/train",
    image_size=(224, 224),
    label_mode="binary",
    batch_size=300,
    shuffle=True
)

# テストデータ画像読込
test_dataset = tf.keras.preprocessing.image_dataset_from_directory(
    "dog_cat_photos/test",
    image_size=(224, 224),
    label_mode="binary",
    batch_size=100,
    shuffle=False
)

# 画像を水増しして訓練データに追加
train_dataset = train_dataset.concatenate(train_dataset.map(flip_left_right))
train_dataset = train_dataset.concatenate(train_dataset.map(rot180))

# データシャッフル
train_dataset = train_dataset.shuffle(300)

# モデル作成
input_layer = tf.keras.Input(shape=(224, 224, 3))
l_layer = tf.keras.applications.mobilenet_v2.preprocess_input(input_layer)
base_model = tf.keras.applications.mobilenet_v2.MobileNetV2(
    input_shape=(224, 224, 3),
    input_tensor=l_layer,
    include_top=False,
    weights="imagenet",
    pooling='avg'
)
base_model.trainable = False
output_layer = tf.keras.layers.Dense(1, activation='sigmoid')
model = tf.keras.Sequential([
    base_model,
    output_layer
])

# コンパイル
model.compile(optimizer="adam", loss='binary_crossentropy', metrics=["accuracy"])

# 学習
model.fit(train_dataset, epochs=20)

# 分類の実行
pred_data = model.predict(test_dataset)

# 評価
model.evaluate(test_dataset)
```
